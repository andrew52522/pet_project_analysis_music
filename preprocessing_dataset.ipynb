```python
import ast

# Convert string representation of list to actual list, handle NaNs
def parse_genres(genres_str):
    if pd.isna(genres_str):
        return []
    try:
        return ast.literal_eval(genres_str)
    except (ValueError, SyntaxError):
        return []

df['parsed_genres'] = df['genres_list'].apply(parse_genres)

# Flatten the list of lists into a single list of all genres
all_genres = [genre for sublist in df['parsed_genres'] for genre in sublist]

# Count the occurrences of each genre
genre_counts = pd.Series(all_genres).value_counts()

print("Occurrences of each genre:")
display(genre_counts.head(20))
```

```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
import time
from tqdm import tqdm  # Прогресс-бар
from google.colab import userdata
```

```python
try:
    cid = userdata.get('SPOTIFY_CID')      # Проверьте, что имя совпадает
    secret = userdata.get('SPOTIFY_SECRET') # Проверьте, что имя совпадает
except Exception as e:
    print("Ошибка! Не могу найти ключи. Проверьте, включены ли галочки 'Доступ из блокнотов' слева.")
    raise e


# ключи апи
cid = userdata.get('SPOTIFY_CID')
secret = userdata.get('SPOTIFY_SECRET')

# авторизация
client_credentials_manager = SpotifyClientCredentials(client_id=cid, client_secret=secret)
sp = spotipy.Spotify(client_credentials_manager=client_credentials_manager)
```

```python
# тянем функцией названия песен. пакетно по 50, потому что 48к песен

all_ids = df['id'].tolist()

# Сюда будем складывать результаты
found_titles = []
# Размер пачки (Spotify разрешает макс 50)
batch_size = 50

for i in tqdm(range(0, len(all_ids), batch_size)):

  # берем срез 50 штук
  batch_ids = all_ids[i:i+batch_size]
  try:

    # запрос на 50 штук
    results = sp.tracks(batch_ids)
    # разбираем ответ
    for track in results['tracks']:
      if track is not None:
        # найден
        found_titles.append(track['name'])
      else:
        # не найден
        found_titles.append(None)

  except Exception as e:
    print(f"Ошибка в пачке {i}: {e}")
    # Если вся пачка упала
    # заполняем None, чтобы длина списка совпала с длиной DataFrame
    found_titles.extend([None] * len(batch_ids))


  time.sleep(0.1) # чтоб не забанили

df['name_song'] = found_titles
```

```python
def simplify_genre(genre_str):
    # Приводим к нижнему регистру на всякий случай
    g = str(genre_str).lower()

    # Приоритеты (если там и поп и рок, выберем рок)
    if 'rap' in g or 'hip hop' in g or 'trap' in g or 'drill' in g:
        return 'Hip-Hop'
    elif 'metal' in g or 'punk' in g or 'rock' in g or 'grunge' in g or 'alternative' in g:
        return 'Rock'
    elif 'country' in g:
        return 'Country'
    elif 'pop' in g or 'singer' in g or 'songwriter' in g:
        return 'Pop'
    else:
        return 'Other'
```

```python
# применим генералищацию тпов
df['main_genre'] = df['genres_list'].apply(simplify_genre)
# выкинем parsed_genres, т.к. дублирует
df.drop('parsed_genres', axis=1, inplace=True)
```

*Чистим текст песни от стоп слов, т.к. LSA чувствителен"*


```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

# Часто бывает в датасете "verse", "hook", "lyrics" в самом начале или конце
custom_stops = {'verse', 'chorus', 'intro', 'outro', 'bridge', 'lyrics'}
stop_words.update(custom_stops)

def remove_stopwords(text):
    if not isinstance(text, str):
        return ""
    # Текст уже в нижнем регистре, просто сплитим
    words = text.split()
    # Оставляем слова, которых нет в стоп-листе и которые длиннее 2 букв
    filtered = [w for w in words if w not in stop_words and len(w) > 2]
    return " ".join(filtered)

df['clean_text'] = df['lyrics'].apply(remove_stopwords)
```

```python
# сделаем фильтрацию где в cleaned_text есть только текст
# Оставляем только те строки, где clean_text НЕ равен пустой строке
df = df[df['clean_text'] != ""]
# Оставляем только те строки, где clean_text НЕ равен пустой строке
df = df[df['clean_text'] != ""]
# Удаляет строки, где clean_text является NaN
df = df.dropna(subset=['clean_text'])
df = df.reset_index(drop=True)   # Сбрасываем индекс

print(f'after cleaning {df.shape}')
```

# СТОП!! БЕРЕМ БОЛЬШЕ СЭМПЛОВ, ВЫКИДЫВАЕМ КАНТРИ

Сэмплирование
Если "Pop" будет 100к песен, а "Rock" 5к, то Попса задавит всё векторное пространство и анализ не репрезентативный будет

 * Почему?

* Ответ: SVD (LSA) работает на основе дисперсии (разброса данных). Если у тебя 20,000 песен Рока и всего 1,000 Кантри, алгоритм решит, что Рок — это и есть вся вселенная, а Кантри — это просто статистическая погрешность (шум).


```python
# все главные категории
df_clean = df[df['main_genre'] != 'Other']

# 2. Делаем балансировку
# n=1100 — сколько брать, чтобы взять country жанр
# Метод sample сам по себе берет строки в случайном порядке
df_balanced = df_clean.groupby('main_genre').sample(n=6700, random_state=52)
```

```python
# Перемешиваем весь итоговый датасет, чтобы жанры шли не блоками, а вперемешку
df_balanced = df_balanced.sample(frac=1, random_state=42).reset_index(drop=True)

# Проверяем результат
print(df_balanced['main_genre'].value_counts())
```

## Отказываемся от стемминга в пользу лемматиации

```python
from nltk.stem import WordNetLemmatizer

# Скачиваем словарь для лемматизации
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    if not isinstance(text, str):
        return ""
    words = text.split()
    # Лемматизируем каждое слово (приводим к базе)
    # pos='v' означает "глагол" (verb). Это поможет превратить singing -> sing
    # Можно лемматизировать без pos, но с глаголами эффективнее для песен
    lemmatized_words = [lemmatizer.lemmatize(w, pos='v') for w in words]
    return " ".join(lemmatized_words)

# Применяем к нашей сбалансированной таблице
# Используй ту колонку, где уже убраны стоп-слова (clean_text или как ты её назвал)
df_balanced['final_text'] = df_balanced['clean_text'].apply(lemmatize_text)

# Смотрим пример до и после
print("До:", df_balanced['clean_text'].iloc[0][:50])
print("После:", df_balanced['final_text'].iloc[0][:50])
```
